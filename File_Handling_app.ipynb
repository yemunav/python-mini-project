```python
import os
print(os.getcwd())
```

```python
## Create a File: -- using "w" write mode:

def create():
    fname=input("Enter the File name to create:")
    f=open(fname,"w")
    f.close()
    print("File created successfully..")
    
```

```python
## Write the data into the created file: -- using "w" write mode:

def write():
    fname=input("Enter the file name you write into:")
    f=open(fname,"w")
    data=input("Enter the data you want in the created file:")
    f.write(data)
    f.close()
    print("Data written successfully..")
```

```python
## Read the File: --using "r" read mode

def read():
    fname=input("enter the file name you want to read:")
    f=open(fname,"r")
    print("-------File Content is as follows--------")
    print(f.read())
    f.close()
```

```python
## Replace the Data into the existing file -- using "w" write mode

def replace():
    fname=input("Enter the file name to replace the content:")
    f=open(fname,"w")
    data=input("Enter the data you want to replace with:")
    f.write(data)
    f.close()
    print("The File conetent replaced successfully...")
```

```python
## Append the data into the existing file ---using "a" append mode

def append():
    fname=input("Enter the file name you want to append with :")
    f=open(fname,"a")
    data=input("Enter the data to append in to the file :")
    f.write(data)
    f.close()
    print("Data appennded successfully...")
```

```python
## Rename the file name: -- using os module and rename() predefined function

def rename():
    old=input("Enter the current file name:")
    new=input("Enter the new file name:")
    os.rename(old,new)
    print("File renamed successfully...")
```

```python
## Delete the file -- using unlink() predefined os module

def delete():
    fname=input("Enter the file to delete:")
    os.unlink(fname)
    print("File Deleted successfully..")
```

```python
## Show the current working directory  --using getcwd() predefined os module

def address():
    print("The current working directory is:")
    print(os.getcwd())
```

```python
## List all files in current directory --using listdir() predefined os module

def list_files():
    print("List of all files:")
    print(os.listdir())
    print("The number of files available:")
    print(len(os.listdir()))
```

```python
## Change the directory --using chdir() predefined module

def change_dir():
    path=input("Enter the directory path to change to:")
    os.chdir(path)
    print("The directory changed successfully....")
```

```python
##### File Handling menu  --using jump statements

while True:
    print("---------------FILE HANDLING MENU---------------")
    print("1. Create a File")
    print("2. Write Data")
    print("3. Read File")
    print("4. Replace File Content")
    print("5. Append to File")
    print("6. Rename File")
    print("7. Delete File")
    print("8. Show Current Directory")
    print("9. List Files in the Directory")
    print("10. Change the Directory")
    print("11. Exit")

    choice=int(input("Enter your Choice:"))

    if choice==1:
        create()
    elif choice==2:
        write()
    elif choice==3:
        read()
    elif choice==4:
        replace()
    elif choice==5:
        append()
    elif choice==6:
        rename()
    elif choice==7:
        delete()
    elif choice==8:
        address()
    elif choice==9:
        list_files()
    elif choice==10:
        change_dir()
    elif choice==11:
        print("Thank you .. Exiting the File handling app..")
        break
    else:
        print("Invalid... Please enter the number between 1 - 11 :")
    
```
